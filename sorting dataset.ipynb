```python
import os
import shutil
```

```python

# for sorting train dataset
source_folder = r"E:\archive\Crema"
destination_folder = os.path.join(source_folder, "Sorted")

emotion_map = {
    "ANG": "Angry",
    "SAD": "Sad",
    "HAP": "Happy",
    "NEU": "Neutral",
    "FEA": "Fear",
    "DIS": "Disgust"
}

# 📁 Step 4: Create folders for each emotion
os.makedirs(destination_folder, exist_ok=True)

for emotion_code, emotion_name in emotion_map.items():
    emotion_folder = os.path.join(destination_folder, emotion_name)
    os.makedirs(emotion_folder, exist_ok=True)

# 📁 Step 5: Move files into respective folders
files_moved = 0

for file_name in os.listdir(source_folder):
    if file_name.endswith(".wav") or file_name.endswith(".mp3"):
        for emotion_code, emotion_name in emotion_map.items():
            if f"_{emotion_code}_" in file_name:
                src = os.path.join(source_folder, file_name)
                dst = os.path.join(destination_folder, emotion_name, file_name)
                shutil.copy2(src, dst)  # Use move() to remove original file
                files_moved += 1
                break

print(f"✅ Done! {files_moved} files sorted into folders by emotion.")
```

```python
#for sorting test

source_folder = r"E:\archive\Savee"  # 🔁 Replace with your folder path
destination_folder = os.path.join(source_folder, "Sorted")

# 📁 Step 2: Emotion code mapping
emotion_map = {
    "a": "Angry",
    "d": "Disgust",
    "f": "Fear",
    "h": "Happy",
    "n": "Neutral",
    "sa": "Sad",
    "su": "Surprise"
}

# 📁 Step 3: Create destination folders
os.makedirs(destination_folder, exist_ok=True)

for emotion_code, emotion_name in emotion_map.items():
    emotion_folder = os.path.join(destination_folder, emotion_name)
    os.makedirs(emotion_folder, exist_ok=True)

# 📁 Step 4: Sort files into corresponding emotion folders
files_moved = 0

for file_name in os.listdir(source_folder):
    if file_name.endswith(".wav") or file_name.endswith(".mp3"):
        lower_name = file_name.lower()
        
        # Check for two-letter codes first (e.g. 'sa', 'su')
        matched = False
        for emotion_code in ["sa", "su"]:
            if f"_{emotion_code}" in lower_name:
                emotion_name = emotion_map[emotion_code]
                src = os.path.join(source_folder, file_name)
                dst = os.path.join(destination_folder, emotion_name, file_name)
                shutil.copy2(src, dst)
                files_moved += 1
                matched = True
                break

        # If not matched already, check single-letter emotions
        if not matched:
            for emotion_code in ["a", "d", "f", "h", "n"]:
                if f"_{emotion_code}" in lower_name:
                    emotion_name = emotion_map[emotion_code]
                    src = os.path.join(source_folder, file_name)
                    dst = os.path.join(destination_folder, emotion_name, file_name)
                    shutil.copy2(src, dst)
                    files_moved += 1
                    break

print(f"✅ Done! {files_moved} files sorted into folders by emotion.")
```

```python
# removing biasness 
target_folder = r"E:\Emotion voice recognition\train_sorted\Angry"  # 🔁 Replace with your actual path

# 🧹 Count of files to delete
files_to_delete = 184

# 📦 Supported audio formats
audio_extensions = [".wav", ".mp3"]

# 📂 List all audio files in the folder
audio_files = [f for f in os.listdir(target_folder)
               if os.path.isfile(os.path.join(target_folder, f)) and f.lower().endswith(tuple(audio_extensions))]

deleted_count = 0

for file_name in audio_files[:files_to_delete]:
    file_path = os.path.join(target_folder, file_name)
    try:
        os.remove(file_path)
        deleted_count += 1
    except Exception as e:
        print(f"Error deleting {file_name}: {e}")

print(f"✅ Deleted {deleted_count} audio files from: {target_folder}")
```
